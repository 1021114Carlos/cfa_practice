*Portfolio Management*

```python
# Investor has the Utility function.
# U = E(R) - 0.5*(A)*Sigma^2

# A = risk aversion coefficient
a = 4

# sigma^2: variance of returns
# Utility score for portfolio X?

expected_return = 12/100
standard_deviation = 20/100

u = expected_return - (0.5*a*standard_deviation**2)
print(f"The Utility score for this portfolio is {u*100:.2f}%")
```
