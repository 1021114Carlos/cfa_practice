```python
# Risk-averse investor with aversion coefficient of 4
ra = 4
er = np.array([0.12, 0.15, 0.21, 0.24])
sd = np.array([0.30, 0.35, 0.40, 0.45])


u_score_4 = er - ((1/2)*ra*(sd**2))
print(f"Utility of investments {u_score_4}")

best_investment = np.argmax(u_score_4) + 1
print("Best investment", best_investment)
```
