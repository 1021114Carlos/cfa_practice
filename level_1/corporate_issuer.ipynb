```python
operating_expense = 70
interest_expense = 15
# 20% increase or decrese in revenue
revenue = 100

# equity-financed  100%
net_income = revenue  - (operating_expense + 20)
roe_financed = net_income/100

# Debt-and-equity 25% equity
debt_equity = revenue  - (operating_expense + interest_expense + 20)
roe_debt_equity = debt_equity/25
print(f"total finance net Income {net_income:.4f}, return on equity {(roe_financed)*100:.4f}\n debt & equity: net income {debt_equity:.4f}, return on equity {(roe_debt_equity)*100:.4f}")

```
