default risk premium

Holding Period Return

```python
# I = Income received
display(Math(r"Holding \ Period \ Return = \frac{(P_{t+1} - P_{t}) + I_{t+1}}{P_{t}}"))
print("\n")

# For longer HPR. R_t -> Annual Returns
display(Math(r"Holding \ Period \ Return = [(1 + R_1)(1 + R_2)(1 + R_3)] - 1"))
```

Arithmetic or Mean Return

```python
display(Math(r"\bar{R}_i = \frac{R_{i,1} + R_{i,2}+...+{R_{i,t-1}} + R_{i,t}}{T} = \frac{1}{T}\sum_{t=1}^{t} {R_{i,t}}"))
```

Geometric Mean Return

```python
display(Math(r"\bar{R}_{g,i} = \sqrt[T]{(1 + R_{i, 1})(1 + R_{i, 2})x...x(1 + R_{i, {t-1}})(1 + R_{i,t})} -1"))
print("\n")
display(Math(r'= \sqrt[T]{\prod_{t=1}^{T}(1 + R_t)} - 1'))
```
